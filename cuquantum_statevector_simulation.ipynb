```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# 1. Create a 10-qubit Entangled Quantum Circuit
qc = QuantumCircuit(10)
qc.h(0)
for i in range(9):
    qc.cx(i, i + 1)
qc.measure_all()

# 2. Run directly on the Tesla T4 GPU via Qiskit Aer GPU (accelerated by cuQuantum)
sim_gpu = AerSimulator(method='statevector', device='GPU')
job = sim_gpu.run(qc)
result = job.result()

print("✓ Success! Circuit executed using NVIDIA cuQuantum GPU backend.")
print("Measurement Results:", result.get_counts())
```
